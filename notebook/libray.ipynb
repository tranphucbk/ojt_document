```python
from autoPrepoccessor3 import AutoPreprocessor
import pandas as pd 
ap = AutoPreprocessor(
    cat_threshold=20,
    one_hot_threshold=10,
    max_missing_ratio=0.7,      # nếu cột missing > 70% -> drop
    add_missing_flags=True,     # thêm cột flag
    missing_flag_suffix="_miss" # ví dụ: "age_miss"
)
df = pd.read_csv("../input/ieee-fraud-detection/train_transaction.csv")
X = ap.fit_transform(df)

print("Cột bị drop:", ap.dropped_columns_)
print("Cấu hình cột:")
print(ap.get_column_configs().head())

print("Dữ liệu sau khi transform:")
print(X.head())
```

```python
ap.print_categorical_columns()
ap.print_fill_values()
```
